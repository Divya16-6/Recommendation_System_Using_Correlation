## Recommendation System Using Correlation

### Dataset Preparation

```python
import pandas as pd

anime = pd.read_csv('anime.csv')
```

### Clean the data

```python
anime.info()
```

```python
anime.isnull().sum()
```

```python
# Here we able to see the null values, need to handle it properly 
```

```python
anime['genre'] = anime['genre'].fillna('Unknown') # Filled genre missed value as Unknown
```

```python
anime['type'] = anime['type'].fillna('Unknown') # Filled type missed value as Unknown
```

```python
anime['rating'] = anime['rating'].fillna(anime['rating'].median()) # Filled rating with the median value of rating
```

```python
anime['episodes'] = pd.to_numeric(anime['episodes'], errors='coerce').fillna(0) # Chaning the episode datatype from object into the numeric value
```

```python
ratings = ratings[ratings.rating != -1]
```

```python
# Merging the anime and rating to make a new dataframe
df = pd.merge(ratings, anime[['anime_id', 'name']], left_on='anime_id', right_on='anime_id')
```

```python
df.head(20)
```

### Recommedation

```python
# Making pivot table to map the anime ID and user ID appropriately

recommend = df.pivot_table(index='user_id', columns='anime_id', values='rating')
```

```python
recommend.head()
```

```python
"""
    Using this function we are finding the correlation of one anime with another

    Args: df --> DataFrame, anime_id --> Based on the anime_id the user watch will be providing the suggestion

    Return: Top 5 anime suggestion
"""
def SuggestionOfAnime(df, anime_id):
    if anime_id not in df.columns:
        return f"Data not found for this Anime ID: {anime_id}"
    similar_anime = df.corrwith(df[anime_id], axis=0)
    similar_anime = pd.DataFrame(similar_anime, columns=['Correlation'])
    similar_anime.dropna(inplace=True)
    similar_anime = similar_anime.sort_values(by='Correlation', ascending=False)
    return similar_anime.head()
```
